```python
x=df.drop('label',axis=1)
y=df['label']
```

```python
x.info()
```

```python
y.info()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=1,test_size=0.2)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_predict=model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
logi_reg_acc=accuracy_score(y_test,y_predict)
print("logistic accuracy is:"+ str(logi_reg_acc))
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model2=DecisionTreeClassifier()
model2.fit(x_train,y_train)
y_predict2=model2.predict(x_test)
```

```python
D_acc=accuracy_score(y_test,y_predict2)
print("Decision tree accuracy is:"+str(D_acc))
```

```python
from sklearn.ensemble import RandomForestClassifier
model3=RandomForestClassifier()
model3.fit(x_train,y_train)
y_pred4=model3.predict(x_test)
```

```python
R_acc=accuracy_score(y_test,y_pred4)
print("Random forest tree is:"+ str(R_acc))
```
